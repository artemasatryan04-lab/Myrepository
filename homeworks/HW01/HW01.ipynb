```python
# Импорт NumPy с общепринятым сокращением
import numpy as np
```

```python
# 1) Создание массива из обычного списка
a = np.array([1, 2, 3])   # одномерный массив
print("a:", a)

# 2) Создание с использованием arange
b = np.arange(0, 10, 2)   # [0, 2, 4, 6, 8]
print("b:", b)

# 3) Массивы zeros и ones
z = np.zeros((3,))        # вектор из нулей длиной 3
o = np.ones((2, 4))       # 2x4 матрица из единиц
print("z:", z)
print("o:\n", o)
```

```python
# 2D массив — из вложенных списков
arr2 = np.array([[1, 2, 3],
                 [4, 5, 6]])
print("arr2:\n", arr2)
```

```python

# Демонстрация атрибутов
print("arr2.shape:", arr2.shape)   # (2, 3) — 2 строки, 3 столбца
print("arr2.dtype:", arr2.dtype)   # тип данных (обычно int64 или int32)
```

```python
# Получение отдельного элемента (строка 0, столбец 1)
elem = arr2[0, 1]
print("arr2[0,1]:", elem)  # должно быть 2

# Извлечение подмассива: первая колонка
col0 = arr2[:, 0]
print("arr2[:,0]:", col0)

# Срез по строкам и столбцам: взять все строки и последние два столбца
sub = arr2[:, 1:]
print("arr2[:,1:]:\n", sub)
```

```python
# поэлементное сложение и умножение
x = np.array([1, 2, 3])
y = np.array([4, 5, 6])

sum_xy = x + y
prod_xy = x * y  # поэлементное умножение

print("x + y =", sum_xy)    # [5, 7, 9]
print("x * y =", prod_xy)   # [4, 10, 18]
```

```python
# среднее, сумма, стандартное отклонение
print("x.mean() =", x.mean())            # 2.0
print("y.mean() =", np.mean(y))          # 5.0
print("arr2.sum() =", arr2.sum())        # сумма всех элементов 1..6 = 21
print("arr2.mean(axis=0) =", arr2.mean(axis=0))  # среднее по столбцам
```
